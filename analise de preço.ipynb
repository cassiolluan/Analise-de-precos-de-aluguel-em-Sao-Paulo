```python
# Calcular o preço médio dos aluguéis
media_aluguel = df_data['Aluguel'].mean()

# Calcular o preço mediano dos aluguéis
mediana_aluguel = df_data['Aluguel'].median()

print(f'Preço médio: R$ {media_aluguel:,.2f}')
print(f'Preço mediano: R$ {mediana_aluguel:,.2f}')
```

Média e mediana dos aluguéis

O preço médio dos aluguéis foi de R$ 3.250,81, enquanto o preço mediano foi de R$ 2.415,00.

A diferença entre os dois valores indica que alguns imóveis com preços de aluguel mais elevados influenciam a média, fazendo com que ela fique muito acima da mediana. A mediana é menos sensível a valores extremos e, neste caso, representa melhor o valor central observado na distribuição dos aluguéis.

```python
# Verificar a quantidade de imóveis por distrito
df_data['Distrito'].value_counts()
```

```python
# Calcular o preço médio do aluguel por distrito
media_por_distrito = df_data.groupby('Distrito')['Aluguel'].mean()

media_por_distrito
```

```python
# Contar quantos imóveis existem em cada distrito
qtd_por_distrito = df_data['Distrito'].value_counts()

# Selecionar apenas distritos com pelo menos 30 imóveis
distritos_validos = qtd_por_distrito[qtd_por_distrito >= 30].index

# Calcular a média de aluguel apenas nesses distritos
media_distritos_validos = (
    df_data[df_data['Distrito'].isin(distritos_validos)]
    .groupby('Distrito')['Aluguel']
    .mean()
)

media_distritos_validos
```

```python
# 5 distritos com maiores preços médios
maiores_medias = media_distritos_validos.sort_values(ascending=False).head(5)

# 5 distritos com menores preços médios
menores_medias = media_distritos_validos.sort_values().head(5)

print("5 maiores médias:")
display(maiores_medias)

print("5 menores médias:")
display(menores_medias)
```

```python
# Relação entre área e preço do aluguel
df_area = df_data[df_data['Área'] > 0]

correlacao_area = df_area['Área'].corr(df_area['Aluguel'])

print(f'Correlação entre área e aluguel: {correlacao_area:.2f}')
```

```python
# Relação entre quantidade de quartos e preço do aluguel
correlacao_quartos = df_data['Quartos'].corr(df_data['Aluguel'])

print(f'Correlação entre quartos e aluguel: {correlacao_quartos:.2f}')
```

```python
# Relação entre quantidade de garagens e preço do aluguel
correlacao_garagem = df_data['Garagem'].corr(df_data['Aluguel'])

print(f'Correlação entre garagens e aluguel: {correlacao_garagem:.2f}')
```

```python
# Visualização da relação entre área e preço do aluguel
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.scatter(
    df_area['Área'],
    df_area['Aluguel'],
    alpha=0.4
)

plt.title('Relação entre Área do Imóvel e Preço do Aluguel')
plt.xlabel('Área (m²)')
plt.ylabel('Aluguel (R$)')

plt.show()
```

O gráfico mostra uma tendência positiva entre a área do imóvel e o preço do aluguel. De modo geral, imóveis com áreas maiores apresentam valores de aluguel mais elevados, embora existam variações entre os imóveis.

```python
# Visualização da relação entre quantidade de quartos e preço do aluguel
plt.figure(figsize=(10, 6))

plt.scatter(
    df_data['Quartos'],
    df_data['Aluguel'],
    alpha=0.4
)

plt.title('Relação entre Número de Quartos e Preço do Aluguel')
plt.xlabel('Número de Quartos')
plt.ylabel('Aluguel (R$)')

plt.show()
```

O gráfico mostra uma relação positiva entre a quantidade de quartos e o preço do aluguel. Imóveis com mais quartos tendem a apresentar valores maiores, embora exista bastante variação dentro de cada quantidade de quartos.

```python
plt.figure(figsize=(10, 6))

maiores_medias.sort_values().plot(kind='barh')

plt.title('Distritos com Maiores Preços Médios de Aluguel')
plt.xlabel('Aluguel médio (R$)')
plt.ylabel('Distrito')

plt.show()
```

O gráfico apresenta os cinco distritos com maiores preços médios de aluguel entre aqueles que possuem pelo menos 30 registros. Jardim Leonor apresenta a maior média entre os distritos selecionados.

## Conclusões

3. A área apresentou a maior relação com o aluguel entre as características analisadas, com correlação de 0,67.

5. A quantidade de quartos apresentou correlação de 0,53, também indicando uma relação positiva, porém menor que área e garagem.
6. Entre os distritos com pelo menos 30 registros, Jardim Leonor apresentou a maior média de aluguel, enquanto Taboão apresentou a menor.

## Limitações da análise

1. Preço anunciado: os valores analisados representam os preços de aluguel anunciados na base, e não necessariamente os valores finais negociados entre proprietário e locatário.

2. Concentração dos anúncios: a quantidade de imóveis varia bastante entre os distritos. Por isso, a comparação entre localidades foi feita considerando apenas distritos com pelo menos 30 registros. 

3. Informações disponíveis: a análise está limitada às características presentes no dataset, como área, quartos, garagem e localização. Outras características que podem influenciar o aluguel não estão disponíveis na base.

4. Registros inconsistentes de área: foram encontrados 2 registros com área igual a zero. Eles foram mantidos na base, mas não foram considerados nas análises que dependem da variável Área.

5. Correlação não significa causa: as relações observadas entre aluguel e características dos imóveis não permitem afirmar que uma característica, isoladamente, seja a causa do aumento ou redução do preço.
